```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from pydantic import BaseModel, Field
```

```python
load_dotenv()
llm = init_chat_model(model="gemini-3.5-flash-lite", model_provider="google_genai")
```

```python
class EvaluationSchema(BaseModel):
    feedback : str = Field(description="Detailed feedback for the essay")
    score : float = Field(description="Score out of 10", ge=0, le=100)
    
structured_model = llm.with_structured_output(EvaluationSchema)
print(structured_model)
```

```python
essay = """
# The Ultimate Driving Machine: The Rise of BMW from Aero Engines to Automotive Icon

## Introduction

Bayerische Motoren Werke (BMW) stands today as one of the preeminent symbols of engineering precision, sporting performance, and luxury in the global automotive industry. Recognized worldwide by its iconic circular blue-and-white roundel and distinctive twin-kidney grille, the Bavarian marque has traversed more than a century of profound transformation. From its origins as a World War I aircraft engine manufacturer facing near-extinction to its rebirth as the pioneer of the modern executive sports sedan, the story of BMW is an extraordinary chronicle of technological resilience, strategic reinvention, and uncompromising brand identity.

---

## 1. Aerial Foundations and Early Rebirth (1916–1939)

The roots of BMW trace back to 1916 with the consolidation of Rapp Motorenwerke and Gustav Otto Flugmaschinenfabrik into Bayerische Flugzeugwerke (BFW), officially renamed Bayerische Motoren Werke in 1917. Founded amidst the pressures of the First World War, the nascent company excelled at high-altitude aero-engine design, most notably the legendary BMW IIIa inline-six engine designed by Max Friz. The engine earned critical acclaim for its efficiency and operational altitude ceiling.

Following the armistice and the subsequent Treaty of Versailles in 1919, German companies were strictly forbidden from producing military aircraft engines. Faced with an existential crisis, BMW diversified rapidly, producing railway brakes, industrial motors, and agricultural equipment. In 1923, Chief Designer Max Friz engineered the **BMW R32 motorcycle**, establishing an engineering layout that remains a BMW Motorrad signature over a century later: a horizontally opposed boxer twin engine coupled with shaft drive.

BMW entered the automobile manufacturing realm in 1928 by acquiring Fahrzeugfabrik Eisenach, makers of the Dixi 3/15—a licensed version of Britain's Austin Seven. Rather than remain a purveyor of utilitarian subcompacts, BMW rapidly evolved its portfolio. By 1936, BMW unleashed the **BMW 328**, a lightweight, aerodynamic roadster powered by a 2.0-liter inline-six. Dominating prestigious competitions including the 1940 Mille Miglia, the 328 permanently embedded athletic performance into BMW's corporate DNA.

---

## 2. Post-War Devastation and the Near Collapse (1945–1959)

World War II brought catastrophic devastation to BMW. Its primary automotive production plant in Eisenach fell behind the Iron Curtain under Soviet administration (later becoming EMW), while the heavily bombed Munich headquarters lay in ruins. In the late 1940s, Allied dismantling directives reduced the facility to producing pots, pans, and bicycles.

Motorcycle production resumed in 1948 with the R24, but the automotive market presented a treacherous landscape. In the 1950s, BMW adopted a polarized, commercially flawed model strategy:
- **Ultra-luxury, low-volume grand tourers:** The elegant BMW 501/502 "Baroque Angels" and the breathtaking Albrecht von Goertz-designed BMW 507 roadster (beloved by Elvis Presley) were critical darlings but severe financial drains.
- **Microcars:** The licensed bubble-car **BMW Isetta** kept the factory lights on, yet generated slender profit margins.

By 1959, BMW was hemorrhaging capital. At the tempestuous annual general meeting on December 9, 1959, the board recommended a buyout by Daimler-Benz. In an extraordinary turn of events, minority shareholders, labor union representatives, and independent dealers revolted and defeated the proposal. Industrialist **Herbert Quandt**, convinced of BMW’s potential, substantially increased his shareholding to nearly 50%, providing the financial stability necessary to chart a bold new course.

---

## 3. The Turning Point: The *Neue Klasse* (1961–1975)

Armed with Quandt family backing, BMW committed its resources to a completely fresh product segment: the **Neue Klasse** (New Class), debuted at the 1961 Frankfurt Motor Show as the **BMW 1500**. 

The Neue Klasse was revolutionary:
1. **Unibody Construction:** Stiff, agile, and lightweight.
2. **Modern M10 Engine:** A robust overhead-camshaft inline-four that would go on to power road cars, Formula 2 racers, and even a 1,400-horsepower turbocharged Brabham-BMW Formula 1 car.
3. **Chassis Dynamics:** Fully independent suspension with front MacPherson struts and semi-trailing rear arms, setting the standard for responsive handling.

The success of the 1500, 1800, and 2000 models created the template for the modern sports sedan. This philosophy reached its apex in 1968 with the launch of the **BMW 02 Series**, particularly the iconic **BMW 2002**. Blending daily practicality, compact dimensions, and blistering rear-wheel-drive dynamics, the 2002 captivated American enthusiasts through car magazines like *Car and Driver*, turning BMW into an international performance benchmark.

---

## 4. Solidifying Identity: The Ultimate Driving Machine and the Rise of M (1970s–1990s)

Under the leadership of visionary Chief Executive **Eberhard von Kuenheim** (who led the company from 1970 to 1993), BMW transformed from a precarious national niche manufacturer into a global titan. Von Kuenheim oversaw the construction of the famous "Four-Cylinder" headquarters (*Vierzylinder*) in Munich and reorganized the entire product portfolio into the structured numbering scheme known today:
- **3 Series (1975):** The benchmark compact executive sports sedan.
- **5 Series (1972):** The quintessential executive luxury express.
- **7 Series (1977):** The flagship technological flagship rivaling the Mercedes-Benz S-Class.
- **6 Series (1976):** The grand touring coupe.

Complementing this reorganization was the brilliant marketing slogan coined by the Ammirati & Puris agency in 1974: **"The Ultimate Driving Machine."** The tagline resonated because the hardware substantiated the claim.

In 1972, BMW established **BMW Motorsport GmbH** (now simply BMW M) under Jochen Neerpasch. Initially tasked with track dominance—spawning the mid-engine **BMW M1** supercar in 1978—the division soon applied its motorsport expertise to production road cars. The arrival of the **E28 M5** in 1984, the world's fastest production sedan at the time, and the homologation-special **E30 M3** in 1986 forged a cultural phenomenon. "M" became arguably the most coveted performance letter in the world.

---

## 5. Strategic Expansion, Challenges, and the SAV Revolution (1990s–2000s)

The 1990s brought aggressive internationalization. BMW broke ground on its first full manufacturing plant outside Germany in Spartanburg, South Carolina, in 1992, signaling its intent to dominate the North American market.

However, rapid expansion brought missteps. In 1994, BMW acquired Britain's Rover Group for £800 million. Beset by aging plants, quality issues, and currency headwinds, the venture proved financially disastrous. In 2000, BMW divested Rover and MG, sold Land Rover to Ford, but astutely retained **MINI**. Relaunched in 2001, MINI became an enormous global hit. Concurrently, BMW negotiated the rights to **Rolls-Royce Motor Cars**, constructing a bespoke facility at Goodwood in 2003 to solidify its ownership of the absolute pinnacle of ultra-luxury motoring.

In 1999, BMW anticipated the global shift toward utility vehicles with the launch of the **BMW X5**. Explicitly branded as a "Sports Activity Vehicle" (SAV) rather than an SUV, the X5 eschewed crude truck-based ladder frames in favor of a unibody car platform tuned for autobahn stability. The X5 and its subsequent siblings (X1, X3, X4, X6, X7) transformed Spartanburg into BMW's largest production plant worldwide and redefined the modern luxury family vehicle.

---

## 6. The Digital and Electrified Era: Charting the Future (2010s–Present)

As environmental regulations tightened and digital connectivity took center stage, BMW adapted ahead of many legacy competitors. In 2011, it launched the **BMW i sub-brand**, debuting the groundbreaking carbon-fiber **BMW i3** electric city car and the futuristic **BMW i8** plug-in hybrid sports car.

Today, BMW navigates the transition to carbon neutrality through a balanced, flexible approach:
- **Electrification (BMW i):** High-volume BEVs such as the i4, iX, i5, and i7 combine zero emissions with traditional BMW driving dynamics.
- **Next-Generation Architecture (*Neue Klasse* 2.0):** Commencing in 2025–2026, BMW's next-generation platform revives the storied "Neue Klasse" moniker, introducing 800-volt high-density battery architectures, revolutionary heads-up computing, and circular lifecycle design.
- **Sustained M Dominance:** Blending hybrid efficiency with twin-turbocharged inline-six and V8 powerplants to satisfy passionate drivers worldwide.

---

## Key Milestones in BMW's History

| Year | Milestone | Significance |
| :--- | :--- | :--- |
| **1916** | Founding of BFW / Rapp Motorenwerke | Company establishes roots in aircraft engine manufacture. |
| **1923** | Release of the BMW R32 Motorcycle | First vehicle fully designed by BMW, introducing the boxer engine layout. |
| **1928** | Acquisition of Fahrzeugfabrik Eisenach | First car production via the Dixi 3/15. |
| **1936** | Introduction of the BMW 328 | Solidified pre-war motorsport prestige and sports-car identity. |
| **1959** | Shareholder revolt & Quandt investment | Averted takeover by Daimler-Benz; saved BMW from corporate extinction. |
| **1961** | Launch of the *Neue Klasse* (BMW 1500) | Saved the company financially; invented the modern compact sports sedan. |
| **1972** | Formation of BMW Motorsport GmbH | Laid foundation for legendary M-badged road cars. |
| **1975** | Introduction of the 3 Series (E21) | Created the best-selling premium sports sedan dynasty in automotive history. |
| **1999** | Debut of the BMW X5 (E53) | Created the "Sports Activity Vehicle" segment and transformed corporate profits. |
| **2013** | Introduction of the BMW i3 and i8 | Pioneered mass-production carbon fiber and modern EV technology. |

---

## Conclusion

The rise of BMW from an engine builder restricted by postwar sanctions into one of the world’s most admired automotive institutions is a study in corporate courage and branding clarity. At critical junctures—whether in 1923, 1959, or 1961—BMW chose bold innovation over retreat. By stubbornly adhering to driver engagement, balanced weight distribution, and engineering excellence, BMW built not merely an automobile manufacturer, but a global culture centered around the sheer joy of driving.
"""
```
