```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BMIState(TypedDict):
    height_m : float
    weight_kg : float
    bmi : float
    level : str
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    height = state["height_m"]
    weight = state["weight_kg"]

    bmi = weight/(height**2)

    state["bmi"] = round(bmi, 2)

    return state

def print_bmi_levels(state: BMIState) -> BMIState:
    bmi = state.get("bmi", 0)

    if bmi < 18.5:
        state["level"] = "Underweight"
    elif 18.5 <= bmi < 25.0:
        state["level"] = "Normal weight"
    elif 25.0 <= bmi < 30.0:
        state["level"] = "Overweight"
    else:
        state["level"] = "Obese"

    return state
```

```python
graph = StateGraph(BMIState)

graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('print_bmi_levels', print_bmi_levels)

graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'print_bmi_levels')
graph.add_edge('print_bmi_levels', END)

workflow = graph.compile()
```

```python
initial_state = {'height_m': 1.7018, 'weight_kg' : 97}

output_state = workflow.invoke(initial_state)

print(output_state)
```
